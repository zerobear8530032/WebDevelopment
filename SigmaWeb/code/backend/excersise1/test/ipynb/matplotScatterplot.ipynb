```python
from matplotlib import pyplot as plt
```

```python
x=[2,3,4,3,5,2,3,4,3,4,2,3,4,3]
y=[2,6,2,4,2,4,3,4,6,5,5,3,4,4]
c=[2,3,4,2,2,3,4,2,3,4,3,4,2,4]
```

```python

plt.scatter(x,y,c=c,cmap="Reds",edgecolors="black",linewidths=1,s=100)
plt.title("scatter plot")

plt.xlabel("x label")
plt.ylabel("y label")
cbar=plt.colorbar()
cbar.set_label("darkness")
plt.show()
```
