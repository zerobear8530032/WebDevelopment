```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
plt.style.available
```

```python
plt.style.use("tableau-colorblind10")
```

```python
slices=(20,15,30,30,5)
lables=["home expenses","study","electicity","taxes","other stuffs"]
```

```python
colors=["blue","red","brown","orange","pink"]
explode=[0,0,0,0.5,0]
plt.pie(slices,shadow=True,labels=lables,autopct="%1.1f%%",colors=colors,explode=explode)
plt.title("pie chart of monthly expenses")
plt.show()
```

```python
colors=["blue","red","brown","orange","pink"]
explode=[0,0,0,0.5,0]
plt.pie(slices,shadow=True,labels=lables,autopct="%1.1f%%",colors=colors,explode=explode,wedgeprops={"edgecolor":"black"})
plt.title("pie chart of monthly expenses")
plt.show()
```
