```python
from matplotlib import pyplot as plt
import numpy as np
axis=plt.axes(projection="3d")
x=np.arange(-2,2,0.01)
y=np.arange(-2,2,0.01)
meshx,meshy=np.meshgrid(x,y)
# meshz=meshx+meshy # create a hyper plane like graph
# meshz=meshx-meshy # create a small hyper plane like graph
# meshz=meshx*meshy # create a curve hyper plane like graph
# meshz=np.cos(meshx)+np.cos(meshy) # create a hill  plane like graph
meshz=np.cos(meshx)*np.cos(meshy) # create wavey hill  plane like graph
axis.plot_surface(meshx,meshy,meshz)
plt.show()
```

```python
# top view
axis=plt.axes(projection="3d")
meshz=np.cos(meshx)*np.cos(meshy) # create wavey hill  plane like graph
axis.plot_surface(meshx,meshy,meshz,cmap="coolwarm")
axis.view_init(azim=5,elev=60)
plt.show()
```

```python
# top view
axis=plt.axes(projection="3d")
meshz=np.cos(meshx)*np.cos(meshy) # create wavey hill  plane like graph
axis.plot_surface(meshx,meshy,meshz,cmap="rainbow")
axis.view_init(azim=11,elev=100)
plt.show()
```
